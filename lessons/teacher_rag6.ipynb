# 🤔 Часть 1. Почему обычной LLM недостаточно?

Большие языковые модели обладают серьёзными ограничениями:

- знания устаревают;
- модель не знает ваши документы;
- возможны галлюцинации.

Именно поэтому появилась архитектура RAG.


# 💡 Что такое RAG?

**Retrieval-Augmented Generation** — подход, при котором языковая модель получает дополнительную информацию из внешней базы знаний перед генерацией ответа.

```text
Вопрос пользователя
        ↓
Поиск информации
        ↓
Найденные документы
        ↓
LLM
        ↓
Ответ
```


# 🏗️ Часть 2. Архитектура RAG

```text
Документы
    ↓
Чанкинг
    ↓
Эмбеддинги
    ↓
Векторная база
    ↓

Запрос пользователя
    ↓
Эмбеддинг запроса
    ↓
Поиск похожих чанков
    ↓
Контекст
    ↓
LLM
    ↓
Ответ
```


# 📂 Этап 1. Загрузка документов

Любая RAG-система начинается с данных.

Источниками могут быть:

- PDF-файлы;
- статьи;
- инструкции;
- базы знаний;
- учебники;
- веб-сайты.

Например:

```text
Правила школы.pdf
Инструкция по олимпиаде.pdf
Положение о пропусках.pdf
```

---

# ✂️ Этап 2. Чанкинг

Большие документы нельзя отправлять в модель целиком.

Поэтому они разбиваются на небольшие части.

Пример:

```text
Документ

↓

Чанк 1
Чанк 2
Чанк 3
Чанк 4
```

Каждый чанк содержит небольшой фрагмент текста.

---

# 🔢 Этап 3. Создание эмбеддингов

Каждый чанк преобразуется в вектор.

Например:

```text
Чанк

↓

[0.12, -0.44, 0.87, ...]
```

Теперь текст можно сравнивать математически.

---

# 🗄️ Этап 4. Векторное хранилище

Все эмбеддинги сохраняются в специальную базу данных.

Например:

- FAISS;
- Chroma;
- Pinecone;
- Weaviate.

На нашем курсе мы будем использовать FAISS.

---

# ❓ Этап 5. Запрос пользователя

Пользователь задаёт вопрос.

Например:

> Сколько длится школьная олимпиада?

---

# 🔢 Этап 6. Эмбеддинг запроса

Вопрос также переводится в вектор.

```text
Вопрос

↓

[0.21, -0.56, 0.73, ...]
```

Теперь запрос и документы находятся в одном векторном пространстве.

---

# 🔍 Этап 7. Поиск похожих документов

Система ищет чанки с наиболее похожими эмбеддингами.

Например:

```text
Запрос

↓

Чанк №17
Чанк №43
Чанк №8
```

Это называется **retrieval**.

---
# 📚 Этап 8. Формирование контекста

Найденные чанки объединяются.

Получается контекст:

```text
Контекст:

Чанк №17

Чанк №43

Чанк №8
```

---

# 🤖 Этап 9. Передача контекста в LLM

Теперь модель получает не только вопрос.

Она получает:

```text
Контекст

+

Вопрос пользователя
```

Например:

```text
Используй информацию ниже.

Контекст:
...

Вопрос:
Сколько длится олимпиада?
```

---

# ✨ Этап 10. Генерация ответа

Теперь модель отвечает, используя найденные документы.

Получается ответ, основанный на данных базы знаний.

---

# 🎯 Почему RAG работает?

Потому что каждая технология выполняет свою задачу.

| Компонент | Роль |
|------------|---------|
| Чанкинг | Делит документы |
| Эмбеддинги | Представляют текст числами |
| FAISS | Выполняет поиск |
| LLM | Формирует ответ |

Вместе они образуют единую интеллектуальную систему.

---


# 🔄 Часть 3. Что такое Pipeline?

До этого момента мы рассматривали отдельные компоненты RAG:

- чанкинг;
- эмбеддинги;
- FAISS;
- LLM.

Но сами по себе эти компоненты бесполезны.

Они начинают приносить пользу только тогда, когда объединяются в единую последовательность действий.

Такую последовательность называют **Pipeline**.

---

## Определение

**Pipeline** — последовательность этапов обработки данных, в которой результат одного этапа становится входом для следующего.

---

Например:

```text
Вопрос пользователя
        ↓
Эмбеддинг запроса
        ↓
Поиск документов
        ↓
Формирование контекста
        ↓
Запрос к LLM
        ↓
Ответ
```

Каждый этап выполняет свою задачу.

Если убрать хотя бы один этап, система перестанет работать корректно.

---

## Почему Pipeline важен?

Представьте автомобильный завод.

Сначала изготавливают детали.

Потом собирают двигатель.

Затем собирают автомобиль.

Каждый этап зависит от предыдущего.

Точно так же устроен RAG.

# 🧩 Часть 4. Роль LangChain

Если реализовывать весь RAG самостоятельно, придётся писать много кода.

Например:

- загрузку документов;
- чанкинг;
- создание эмбеддингов;
- подключение FAISS;
- поиск документов;
- формирование prompt;
- обращение к LLM.

Все эти действия можно выполнять вручную.

Но это долго и неудобно.

---

## Что такое LangChain?

**LangChain** — библиотека для создания приложений на основе больших языковых моделей.

Она предоставляет готовые инструменты для:

- загрузки документов;
- чанкинга;
- работы с эмбеддингами;
- подключения векторных баз;
- создания RAG-систем.

---

## Почему LangChain стал популярным?

Потому что позволяет собирать сложные системы из готовых блоков.

Вместо сотен строк кода можно использовать несколько готовых компонентов.

---

## Аналогия

Представьте конструктор LEGO.

Можно изготовить каждую деталь самостоятельно.

А можно взять готовые детали и быстро собрать модель.

LangChain играет роль такого конструктора для AI-приложений.

# 📝 Часть 5. Из чего состоит код RAG?

Минимальный RAG обычно состоит из пяти этапов.

---

## Этап 1. Загрузка документов

```python
loader = TextLoader("document.txt")
documents = loader.load()
```

Что происходит?

1. Открывается файл.
2. Текст считывается в память.
3. Получается список документов.

---

## Этап 2. Чанкинг

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)
```

Что происходит?

1. Документ разбивается на части.
2. Каждая часть становится отдельным чанком.
3. Чанки будут участвовать в поиске.

---

## Этап 3. Создание эмбеддингов

```python
embeddings = OpenAIEmbeddings()
```

Что происходит?

Модель эмбеддингов преобразует текст в числовые векторы.

---

## Этап 4. Создание FAISS

```python
vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)
```

Что происходит?

1. Для каждого чанка создаётся эмбеддинг.
2. Все эмбеддинги сохраняются в FAISS.
3. База становится готовой к поиску.

---

## Этап 5. Поиск

```python
docs = vectorstore.similarity_search(question)
```

Что происходит?

1. Вопрос преобразуется в эмбеддинг.
2. Выполняется поиск похожих чанков.
3. Возвращаются наиболее подходящие документы.

# 🎯 Часть 6. Как устроен Prompt в RAG?

Многие думают, что найденные документы автоматически понимаются моделью.

На самом деле нет.

Документы необходимо явно передать в запрос.

---

## Пример плохого запроса

```text
Вопрос:
Когда проходит олимпиада?
```

Модель не имеет доступа к найденным документам.

---

## Пример правильного запроса

```text
Используй информацию ниже для ответа.

Контекст:

Олимпиада проводится
с 10 по 15 мая.

Вопрос:

Когда проходит олимпиада?
```

Теперь модель может использовать найденный контекст.

---

## Что обычно входит в Prompt?

Большинство RAG-промптов состоит из трёх частей:

```text
Инструкция

Контекст

Вопрос пользователя
```

Например:

```text
Ты помощник по школьным документам.

Используй только информацию из контекста.

Контекст:
...

Вопрос:
...
```

---

## Почему Prompt важен?

Даже если поиск работает идеально, плохой prompt может ухудшить качество ответа.

Поэтому Retrieval и Prompt работают вместе.

---

# 🧠 Что важно запомнить?

RAG — это не новая модель.

RAG объединяет:

- поиск информации;
- эмбеддинги;
- векторные базы данных;

Именно так работают современные AI-ассистенты.
