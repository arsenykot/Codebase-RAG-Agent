## 1. Как компьютер видит текст?

Компьютер умеет работать **только с числами**.

Попробуем посмотреть, как компьютер видит слово `Привет`.

````
text = "Привет"

for symbol in text:
    print(symbol, "→", ord(symbol))
````

### Вопрос для размышления 🤔

Почему этого недостаточно?

Компьютер знает:

```text
П → 1055
р → 1088
```

Но он **не понимает смысл** слова.

Для компьютера:

```text
кот
```

и

```text
кошка
```

— это просто разные наборы символов.


# 2. Способы представления текста

## Способ №1 — Символьная токенизация (Character Tokenization)

Текст разбивается на **отдельные символы**.

Пример:

```text
Привет
↓
['П', 'р', 'и', 'в', 'е', 'т']
```


````
text = "Привет, мир!"

tokens = list(text)

print(tokens)
print("Количество токенов:", len(tokens))
````

### Плюсы и минусы

✅ Подходит для любого языка  
✅ Нет неизвестных слов

❌ Очень длинные последовательности  
❌ Медленно для модели  
❌ Плохо понимается смысл


## Способ №2 — Токенизация по словам (Word Tokenization)

Теперь разбиваем текст на **слова**.


````
text = "Я люблю машинное обучение"

tokens = text.split()

print(tokens)
print("Количество токенов:", len(tokens))
````

### Проблема word tokenization

Посмотрите:

```text
кот
кота
коту
котом
```

Для компьютера это **4 разных слова**.

А ещё есть новые слова:

```text
супермегакриптокот
```

Что делать модели, если такого слова раньше не было?


## Способ №3 — Subword Tokenization (современный подход)

Современные модели используют **кусочки слов**.

Например:

```text
машиностроение
↓
['машин', 'остро', 'ение']
```

Или:

```text
суперкот
↓
['супер', 'кот']
```

Так модель может понимать **новые слова**.


# 3. Современные алгоритмы токенизации

## 1. BPE (Byte Pair Encoding)

Используется в GPT-моделях.

### Как работает?

Сначала слово:

```text
к о т
```

Потом модель замечает, что некоторые символы часто встречаются вместе:

```text
ко
```

Затем:

```text
кот
```

Частые части объединяются.

---

## 2. WordPiece

Использовался в BERT.

Работает похоже на BPE, но выбирает части слова более «умно» — через вероятность.

---

## 3. SentencePiece

Может работать **даже без пробелов**.

Подходит для китайского и японского языков.


## Почему сейчас используют Subword?

| Метод | Плюс | Минус |
|---|---|---|
| Символы | Любой язык | Очень длинно |
| Слова | Простота | Не знает новые слова |
| Subword | Баланс | Сложнее |

### Почему победил Subword?

1. Понимает **новые слова**  
2. Требует **меньше памяти**  
3. Быстрее, чем посимвольный подход


# 4. Практика — токенизация как у GPT

````
!pip -q install tiktoken
````

Теперь используем настоящую библиотеку токенизации, которую используют GPT-модели.


````
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4o")

text = "Привет! Я изучаю NLP"

tokens = encoding.encode(text)

print("Токены:", tokens)
print("Количество токенов:", len(tokens))
````

### Посмотрим, как GPT разбивает текст

````
for token in tokens:
    print(token, "→", encoding.decode([token]))
````

## Практическое задание 1 🔬

Проверьте, что занимает больше токенов:

- `машина`
- `automobile`
- `🚀`
- `Привет мир!`

Запишите результаты.


````
texts = [
    "машина",
    "automobile",
    "🚀",
    "Привет мир!"
]

for text in texts:
    tokens = encoding.encode(text)
    print(f"{text}: {len(tokens)} токенов → {tokens}")
````

# 5. Важная идея для RAG

Когда мы строим RAG-систему:

1. Документы разбиваются на текстовые куски  
2. Текст превращается в токены  
3. Модель читает только часть информации

### Ограничение

У моделей есть **контекстное окно**.

Например:

```text
128 000 токенов
```

Нельзя загрузить бесконечно большой текст.

Поэтому в RAG сначала работает **поиск**, а потом уже генерация ответа.
