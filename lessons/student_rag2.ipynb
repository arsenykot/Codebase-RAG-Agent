## Задание 1 — Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "кот любит молоко",
    "собака любит мясо",
    "кот любит рыбу"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

### Попробуйте изменить тексты.

```python
texts = [
    "ваш текст",
    "ещё текст",
    "я ещё текст"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

## Задание 2 — N‑grams

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "я люблю школу",
    "я не люблю школу"
]

vectorizer = CountVectorizer(ngram_range=(1,2))
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

## Задание 3 — TF‑IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    "кот любит молоко",
    "кот любит рыбу",
    "рыба любит воду",
    "кот кот кот"
]

tfidf = TfidfVectorizer()
X = tfidf.fit_transform(docs)

print(tfidf.get_feature_names_out())
print(X.toarray())
```

```python
!pip install pymorphy3

import pymorphy3
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')  # для новых версий NLTK

# Инициализируем морфологический анализатор
morph = pymorphy3.MorphAnalyzer()

def russian_tokenizer(text):
    """Токенизатор с лемматизацией для русского языка"""
    tokens = word_tokenize(text.lower())
    lemmas = []
    for token in tokens:
        if token.isalpha():  # только буквенные токены
            lemma = morph.parse(token)[0].normal_form
            lemmas.append(lemma)
    return lemmas

documents = [
    "Кошки сидели на окнах и смотрели на улицу",
    "Собака бегала по улицам и лаяла на прохожих",
    "Кошкам и собакам нужно хорошее питание",
    "Птицы сидели на ветках деревьев"
]

vectorizer = TfidfVectorizer(
    tokenizer=russian_tokenizer,
    token_pattern=None,
    min_df=1,
    max_df=0.9
)

tfidf_matrix = vectorizer.fit_transform(documents)
print("Лемматизированные признаки:", vectorizer.get_feature_names_out())
print("\nTF-IDF матрица:\n", tfidf_matrix.toarray())
```
