## Задание 3.1. Первый запрос через OpenRouter

**Инструкция:** Вставь свой API-ключ в переменную `OPENROUTER_API_KEY` и выполни код.

```python
import openai

# TO-DO Вставь свой API ключ из OpenRouter
OPENROUTER_API_KEY = ""

# Настройка клиента для OpenRouter
client = openai.OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)

# Отправка запроса
response = client.chat.completions.create(
    model="openrouter/free",
    messages=[
        {"role": "system", "content": "Ты — полезный ассистент."},
        {"role": "user", "content": "Что такое машинное обучение?"}
    ],
    temperature=0.7,
    max_tokens=200,
    # Дополнительные параметры OpenRouter (опционально):
    # headers={
    #     "HTTP-Referer": "http://localhost:3000",  # URL вашего приложения
    #     "X-Title": "My App",  # Название вашего приложения
    # }
)

# Вывод ответа
print(response.choices[0].message.content)
```

## Задание 3.2. Работа с параметрами

**Задача:** Поэкспериментируй с параметром `temperature` и посмотри, как он влияет на ответ.

```python
import requests
import json


headers = {
    "Authorization": f"Bearer {OPENROUTER_API_KEY}",
    "Content-Type": "application/json"
}

prompt = "Придумай 3 названия для стартапа в сфере искусственного интеллекта"

# Разные значения temperature
temperatures = [0.1, 0.7, 1.5]

for temp in temperatures:
    print(f"\n{'='*50}")
    print(f"Temperature = {temp}")
    print('-' * 50)

    payload = {
        "model": "openrouter/free",
        "messages": [{"role": "user", "content": prompt}],
        "temperature": temp,
        "max_tokens": 200
    }

    response = requests.post(
        url="https://openrouter.ai/api/v1/chat/completions",
        headers=headers,
        data=json.dumps(payload)
    )

    if response.status_code == 200:
        result = response.json()
        answer = result["choices"][0]["message"]["content"]
        print(answer)
    else:
        print(f"Ошибка: {response.status_code}")
```
